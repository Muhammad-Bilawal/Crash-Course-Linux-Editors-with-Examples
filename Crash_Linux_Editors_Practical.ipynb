## ✏️ What is a Text Editor?

A text editor is a tool used to create and edit text files.

👉 Used in Linux for:
- Writing code
- Editing configuration files
- Managing system settings


## ⚡ Vi Editor (Powerful but advanced)

### Command to open:
```bash
vi file.txt
```

### 🔑 Basic Commands:
| Action | Command |
|------|--------|
| Insert Mode | i |
| Delete Line | dd |
| Copy (Yank) | yy |
| Paste | p |
| Undo | u |
| Search | /word |
| Save | :w |
| Save & Exit | :wq |


````python
# Practice Activity (Simulation)
print("Type these commands in your Linux terminal:")
print("vi file.txt")
print("Press i to insert text")
print("Press ESC then :wq to save & exit")
````

## 😊 Nano Editor (Beginner Friendly)

### Command to open:
```bash
nano file.txt
```

### 🔑 Basic Commands:
| Action | Shortcut |
|------|--------|
| Save | Ctrl + O |
| Exit | Ctrl + X |
| Search | Ctrl + W |


````python
# Practice Activity
print("Open Nano:")
print("nano file.txt")
print("Write text, then press Ctrl+O to save")
print("Press Ctrl+X to exit")
````

## 🔁 Important Concepts

### 🗑 Delete
- Remove text
- Example: `dd` (delete line)

### 📋 Yank (Copy)
- Copy text
- Example: `yy`

### 📥 Paste
- Paste copied text
- Example: `p`

### ↩ Undo
- Reverse last action
- Example: `u`


## 🔍 Searching

Use:
```bash
/word
```

Press `n` for next result.

## 💾 Saving & Exiting

### Vi Editor:
- Save → `:w`
- Exit → `:q`
- Save & Exit → `:wq`
- Force Exit → `:q!`

### Nano Editor:
- Save → `Ctrl + O`
- Exit → `Ctrl + X`
